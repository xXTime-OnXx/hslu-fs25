# CVAI - Calibration

In this example we are going to experiment with camera calibration.

```python
import numpy as np
import matplotlib.pyplot as pl
import cv2 as cv
```

Let's read an image with a chessboard pattern. You have 3 images to test:

Chessboard1.jpg

Chessboard2.jpg

Chessboard3.jpg

What was varied when these were captured?

```python
img1 = cv.imread ('Images/Chessboard1.jpg')
image_rgb = cv.cvtColor (img1, cv.COLOR_BGR2RGB)
pl.imshow (image_rgb)
```

This image is of very high resolution. Some OpenCV functions won't work with this. We better downsample a few times. You may vary how often.

```python
print("Size of image before pyrDown: ", image_rgb.shape)
image_rgb = cv.pyrDown(image_rgb)
print("Size of image after pyrDown: ", image_rgb.shape)
pl.imshow(image_rgb)
```

```python
print("Size of image before pyrDown: ", image_rgb.shape)
image_rgb = cv.pyrDown(image_rgb)
print("Size of image after pyrDown: ", image_rgb.shape)
pl.imshow(image_rgb)
```

```python
print("Size of image before pyrDown: ", image_rgb.shape)
image_rgb = cv.pyrDown(image_rgb)
print("Size of image after pyrDown: ", image_rgb.shape)
pl.imshow(image_rgb)
```

Now let's extract the (8,6) inner corners of this chessboard.

```python
ret, corners = cv.findChessboardCorners(image_rgb, (8,6), None)
#if ret == True:
#    print(corners)
```

Then let's visualize these corners in the image.

```python
cv.drawChessboardCorners(image_rgb, (8,6), corners, ret)
pl.imshow(image_rgb)
```

Finally, let's estimate the instrinsic and extrinsic calibration parameters, as well as the distortion coefficients. The 3D points in real world space will be initialized appropriately. What can you interpret comparing the results for the 3 images?

```python
# prepare object points, like (0,0,0), (1,0,0), (2,0,0) ....,(6,5,0)
objp = np.zeros((6*8,3), np.float32)
objp[:,:2] = np.mgrid[0:8,0:6].T.reshape(-1,2)

# Arrays to store object points and image points from all the images.
objpoints = [] # 3d point in real world space
imgpoints = [] # 2d points in image plane.

objpoints.append(objp)
imgpoints.append(corners)
#print(objp)
#print(corners)

ret, mtx, dist, rvecs, tvecs = cv.calibrateCamera(objpoints, imgpoints, image_rgb.shape[:2], None, None)

print('Reprojection error', ret)
print('')
print('Intrinsic matrix\n', mtx)
print('\nDistortion coefficients\n', dist)
print('\nRotation\n', rvecs)
print('\nTranslation\n', tvecs)
```
