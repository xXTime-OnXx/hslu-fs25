# Long Short-Term Memory
In this exercise, we will implement an LSTM. In the class, we have already seen the definition of the LSTM update rules at time step $t$:

$$
\begin{align}
f_t &= \sigma(W_f h_{t-1} + U_f x_t + b_f) \\
i_t &= \sigma(W_i h_{t-1} + U_i x_t + b_i) \\
o_t &= \sigma(W_o h_{t-1} + U_o x_t + b_o) \\
\tilde{c}_t &= \tanh(W_c h_{t-1} + U_c x_t + b_c) \\
c_t &= f_t * c_{t-1} + i_t * \tilde{c}_t \\
h_t &= o_t * \tanh(c_t)
\end{align}
$$

```python
import torch
import torch.nn as nn
```

Implement this original version of the LSTM as an `LSTMCell`.

```python
class LSTMCell(nn.Module):
    
    def __init__(self, input_dim, hidden_dim, num_chunks=4):
        super().__init__()
        
        # Forget Gate
        self.W_hf = nn.Parameter(torch.zeros(hidden_dim, hidden_dim))
        self.W_xf = nn.Parameter(torch.zeros(hidden_dim, input_dim))
        self.b_f = nn.Parameter(torch.zeros(hidden_dim))
        
        # Input Gate
        self.W_hi = nn.Parameter(torch.zeros(hidden_dim, hidden_dim))
        self.W_xi = nn.Parameter(torch.zeros(hidden_dim, input_dim))    
        self.b_i = nn.Parameter(torch.zeros(hidden_dim))
        
        # Output Gate
        self.W_ho = nn.Parameter(torch.zeros(hidden_dim, hidden_dim))
        self.W_xo = nn.Parameter(torch.zeros(hidden_dim, input_dim))       
        self.b_o = nn.Parameter(torch.zeros(hidden_dim))
        
        # Cell Memory
        self.W_hc = nn.Parameter(torch.zeros(hidden_dim, hidden_dim))
        self.W_xc = nn.Parameter(torch.zeros(hidden_dim, input_dim))       
        self.b_c = nn.Parameter(torch.zeros(hidden_dim))
        
        # Short writing
        # self.w = nn.Linear(hidden_dim, num_chunks * hidden_dim, bias=False)
        # self.u = nn.Linear(input_dim, num_chunks * hidden_dim, bias=False)
        # self.b = nn.Parameter(torch.empty(num_chunks * hidden_dim))
        
    
    def reset_parameters(self):
        for weight in self.parameters():
            nn.init.normal_(weight, 0, 1)
            
    
    def forward(self, x, hidden_state, cell_state):
        f_t = torch.sigmoid(self.W_hf @ hidden_state + self.W_xf @ x + self.b_f)
        i_t = torch.sigmoid(self.W_hi @ hidden_state + self.W_xi @ x + self.b_i)
        o_t = torch.sigmoid(self.W_ho @ hidden_state + self.W_xo @ x + self.b_o)
        c_memory = torch.tanh(self.W_hc @ hidden_state + self.W_xc @ x + self.b_o)
        
        c_state = f_t * cell_state + i_t * c_memory
        h_state = o_t * torch.tanh(c_state)
        
        return h_state, c_state
```
